# Integers and Floats 

These are the number types, they are fairly simple, the only suprising operators are //, %, **

```python
# examples
print(3+5) # addition
print(2*5) # multiplication
print(5/2) # will return whole number + fraction as decimal point
print(5/5) # / *always* returns a decimal
print(20//7) # whole division
print(20%7) # remainder after whole division (pronounced "modulo" or "mod")
print(2**10) # exponent
```

#### A quick aside about types and the type() function

### *EVERYTHING* in python is an object (i.e. an instance of a class)

For example, the integer class has objects 3, 42, -10, ...

To find out what type (from which class) is a thing, you can use the built in type() function.

```python
print(type(42))
print(type(3.0))
print(type(type))
print(type(print))
```

### A quick aside on "casting"

The word "cast" means to transform something of one class into something of a different class. This works when the objects are similar. If they are too dissimilar, then the cast will fail.

```python
print(float(5))# turn an integer into a float
print(int(3.9)) # integers return the *floor*, i.e. delete everything after .
```

# Strings

Strings are text sequences in python.  A sequence is the python name for an array (what it would be called in data structures).

An array is:
1. Contiguous (there's no breaks)
2. has numbered indices (positions)

```python
print('hello')
print(type('hello'))
```

## definining a string

You can define a string with single quotes '', double quotes "", or triple quotes ''' '''.

```python
print('hello')
print("hello")
print('''hello''')
```

Here are the reasons why so many kinds are supported (basically to break ambiguity and to be lazy when you can)

```python
print('that's mine')
```

```python
print("that's mine")
```

```python
print("he said, "stop!"") # similar problem with double quotes
```

```python
print('he said, "stop!"')
```

```python
print('''he said, "that's mine!"''')
```

## Escape characters

Escape characters are special symbols that help break ambiguity and represent textual things that are not text themselves.

```python
print('he said, "that\'s mine!"') # breaks ambiguity
print('one line\ntwo line') # \n is the new line character
```

### built in string functions

The string type is very rich. It has a lot of complexity. It's  a collection of things (characters in the string vs. e.g. an integer 5 that is a number and that's all).

Because of the above, there are a lot of functions that are built in to the string: https://docs.python.org/3/builtins/stdtypes.html#str 

There are too many to go over in lecture. You don't need to memorize them all, but they're useful to know/use because it can save you a lot of work.

```python
print('hello'.upper())
print('hello'.center(10))
print('hello'.center(10,'-'))
```

#### Arithemetic operators on strings behave differently than for numbers

```python
print('55'+'10')
print('55'*10)
print('55'-'10') # - and / are not supported
```

# Indexing a string

Because a string is an array, you can "index" it, by referencing in brackets [] the numbered position. You will be returned the contents at that position.

```python
'hey, that\'s mine!'[10] # note that the \' is a *single* character
```

```python
'hey, that\'s mine!'[-1] # negative indices start at -1 from the far right
```

```python
# this does the same as above, but ugh, it's so ugly!
'hey, that\'s mine!'[len('hey, that\'s mine!')-1]
```

```python
print(len('hi'))
print('hi'[0])
print('hi'[1])
print('hi'[2])
```

#### How to have integers and strings "play nice" with each other

```python
print("10"*5)  # concatenate the string "10" to itself 5 times
```

```python
print("8573923 * 23 is " + 8573923 * 23)
```

```python
# you can cast to fix, but becomes cumbersome (annoying)
print("8573923 * 23 is " + str(8573923 * 23)) 
```

```python
print("8573923 * 23 is {} and 8573923 * 22 is {}".format(8573923 * 23,8573923 * 22)) 
```

```python
# The beloved f-string
print(f"8573923 * 23 is {8573923 * 23} and 8573923 * 22 is {8573923 * 22}") 
print(f"7/3 is {7/3}") # ahhh, so many decimals!
print(f"7/3 is {7/3:.2}") # let's only go 2 decimals far
```
