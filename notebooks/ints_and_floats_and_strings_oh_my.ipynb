# Integers and Floats 

These are the number types, they are fairly simple, the only suprising operators are //, %, **

```python
# examples
print(3+5) # addition
print(2*5) # multiplication
print(5/2) # will return whole number + fraction as decimal point
print(5/5) # / *always* returns a decimal
print(20//7) # whole division
print(20%7) # remainder after whole division
print(2**10) # exponent
```

#### A quick aside about types and the type() function

### *EVERYTHING* in python is an object (i.e. an instance of a class)

For example, the integer class has objects 3, 42, -10, ...

To find out what type (from which class) is a thing, you can use the built in type() function.

```python
print(type(42))
print(type(3.0))
```

### A quick aside on "casting"

The word "cast" means to transform something of one class into something of a different class. This works when the objects are similar. If they are too dissimilar, then the cast will fail.

```python
print(float(5))# turn an integer into a float
print(int(3.9)) # integers return the *floor*, i.e. delete everything after .
```
